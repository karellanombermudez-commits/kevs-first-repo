# Fake US demographics: map and graph

This notebook creates **synthetic state-level data** for all 50 US states, then maps population and compares median household income. The values are randomly generated for demonstration and do not describe real people or current state statistics.

## Create reproducible fake data

A fixed random seed makes the same example data each time you run the notebook.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

states = {
    "Alabama": "AL", "Alaska": "AK", "Arizona": "AZ", "Arkansas": "AR",
    "California": "CA", "Colorado": "CO", "Connecticut": "CT",
    "Delaware": "DE", "Florida": "FL", "Georgia": "GA",
    "Hawaii": "HI", "Idaho": "ID", "Illinois": "IL",
    "Indiana": "IN", "Iowa": "IA", "Kansas": "KS",
    "Kentucky": "KY", "Louisiana": "LA", "Maine": "ME",
    "Maryland": "MD", "Massachusetts": "MA", "Michigan": "MI",
    "Minnesota": "MN", "Mississippi": "MS", "Missouri": "MO",
    "Montana": "MT", "Nebraska": "NE", "Nevada": "NV",
    "New Hampshire": "NH", "New Jersey": "NJ", "New Mexico": "NM",
    "New York": "NY", "North Carolina": "NC", "North Dakota": "ND",
    "Ohio": "OH", "Oklahoma": "OK", "Oregon": "OR",
    "Pennsylvania": "PA", "Rhode Island": "RI", "South Carolina": "SC",
    "South Dakota": "SD", "Tennessee": "TN", "Texas": "TX",
    "Utah": "UT", "Vermont": "VT", "Virginia": "VA",
    "Washington": "WA", "West Virginia": "WV", "Wisconsin": "WI",
    "Wyoming": "WY",
}

demographics = pd.DataFrame({
    "state": list(states),
    "state_code": list(states.values()),
    "population": rng.integers(600_000, 40_000_000, size=len(states)),
    "median_age": rng.integers(29, 46, size=len(states)),
    "median_household_income": rng.integers(42_000, 105_000, size=len(states)),
})
demographics.head()
```

## Map: synthetic population by state

Hover over a state to see its synthetic population and other example values. The colors show population on a logarithmic scale so smaller states remain visible.

```python
import plotly.express as px

population_map = px.choropleth(
    demographics,
    locations="state_code",
    locationmode="USA-states",
    color="population",
    hover_name="state",
    hover_data={
        "population": ":,",
        "median_age": True,
        "median_household_income": ":,",
        "state_code": False,
    },
    color_continuous_scale="Viridis",
    title="Synthetic Population by State (Fake Data)",
)
population_map.update_geos(scope="usa", showlakes=True, lakecolor="white")
population_map.update_layout(margin={"r": 0, "t": 50, "l": 0, "b": 0})
population_map.show()
```

## Graph: states with the highest synthetic household incomes

```python
top_income = demographics.nlargest(10, "median_household_income").sort_values("median_household_income")
income_chart = px.bar(
    top_income,
    x="median_household_income",
    y="state",
    orientation="h",
    labels={"median_household_income": "Median household income ($)", "state": "State"},
    title="Top 10 Synthetic Median Household Incomes",
    text_auto=",d",
)
income_chart.update_layout(showlegend=False)
income_chart.show()
```

**Reminder:** all demographic values in this notebook are randomly generated examples, not official statistics. The interactive US map requires an internet connection when Plotly retrieves its state boundaries.
