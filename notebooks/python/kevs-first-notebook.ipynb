# Kev's first notebook 🐸

A notebook combines **executable Python**, formatted explanations, and rich results in one document. Run a cell with **Shift+Enter**. Variables remain available to later cells in the same kernel.

## 1. Run code and keep its results nearby

This cell creates a small dataset. Try changing a number and rerunning it.

```python
import numpy as np
import pandas as pd

days = np.arange(1, 8)
temperatures = np.array([18, 20, 19, 23, 25, 22, 21])
weather = pd.DataFrame({"day": days, "temperature_c": temperatures})
weather
```

## 2. Explore data interactively

DataFrames display as readable tables. You can filter or summarize them in a new cell without rerunning the whole notebook.

```python
warm_days = weather[weather["temperature_c"] >= 22]
print(f"Average temperature: {weather['temperature_c'].mean():.1f} °C")
warm_days
```

## 3. Make a quick visualization

This simple text chart needs no extra plotting package. Each block represents one degree above 15 °C.

```python
for row in weather.itertuples(index=False):
    bar = "*" * (row.temperature_c - 15)
    print(f"Day {row.day}: {bar} {row.temperature_c} °C")
```
